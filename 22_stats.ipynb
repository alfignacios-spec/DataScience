```python
import pandas as pd
import numpy as np
array=np.array([2,1,5,6,4,5,2,1,6,6,6,6,3,3,3,2]),
series=pd.Series(array)
```

```python
#IQR
q1=np.percentile(series,25)
print(q1)

q3=np.percentile(series,75)
print(q3)
```
