Grafico de lineas
Se usa cuando el eje X sigue un orden temporal

```python
from matplotlib import pyplot as plt
```

```python
x=[1,2,3,4,5]
y=[10,20,25,30,45]
plt.plot(x,y, marker='o',linestyle='-',color='r')
plt.xlabel('Periodo')
plt.ylabel('Valor')
plt.title('Grafico de lineas')

plt.show()
```

grafico de barras

```python
categorias= ['A', 'B','C']
valores=[12,34,67]
plt.bar(categorias,valores,color='#F9C424')
plt.xlabel('categorias')
plt.ylabel('valores')
plt.title('Grafico de barras')

plt.show()
```

```python
categorias= ['A', 'B','C','D','E']
valores=[12,34,67,15,54]
plt.barh(categorias,valores,color='#F9C424')
plt.xlabel('categorias')
plt.ylabel('valores')
plt.title('Grafico de barras')

plt.show()
```

```python
frutas=['manzana','pera','durazno']
porcentajes=[50,20,30]
plt.pie(porcentajes,labels=frutas,autopct='%1.1f%%')
plt.title('grafico de pastel')
plt.show()
```

Grafico de dispersion (scatter plot)

```python
import numpy as np
x=np.random.rand(50)
y=np.random.rand(50)

plt.scatter(x,y,color='blue')
plt.xlabel('variable x')
plt.ylabel('variable y')

plt.title('Grafico de dispersion')
plt.show()
```

```python
datos=np.random.randn(1000)
plt.hist(datos,bins=25, color='blue',edgecolor='black')
plt.xlabel('valor')
plt.ylabel('frecuencia')
plt.title('Histograma')
plt.show()
```

```python
data_box=[np.random.normal(0,std,100) for std in range (1,4)]
plt.boxplot(data_box,orientation='vertical', patch_artist='True')
plt.xlabel('Grupo')
plt.ylabel('Valor')
plt.title('Grafico Boxplot')
plt.show()
```
