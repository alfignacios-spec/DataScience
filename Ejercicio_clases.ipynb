```python
#Ejercicio 1
class Persona:
    def __init__(self,nombre:str,edad:int)->None: #Siempre va self, creamos los parametros (elementos) nombre y edad
        self.nombre=nombre
        self.edad=edad

    def saludar(self):
        print(f'Hola soy {self.nombre} y tengo {self.edad} años')

jorge=Persona('Jorge',10) #creamos objetos variados
maria=Persona('Maria',20)
luis=Persona('Luis',15)

jorge.saludar()
maria.saludar()
luis.saludar()
```

```python
#Ejercicio 2
class Rectangulo:
    def __init__(self,base:float, altura:float)->None: #definimos una metodo iniciador
        self.base=base #definimos atributos
        self.altura=altura

    def area(self)->float: #definimos otros metodos, deben tener siempre la misma estructura de self
        return self.base*self.altura
    
    def perimetro(self)->float:
        return 2*(self.base+self.altura) #return porque aun trabajremos con este dato, no queremos solamente visualizarlo
    
    def __str__(self)->str: #el metodo str te devuelve un print
        print(f'Rectangulo base: {self.base} y altura: {self.altura}') #print, porque solo queremos un mensaje de salida, no utilizaremos mas este dato
        
mi_rectangulo=Rectangulo(3,5)
print(mi_rectangulo.area())
print(mi_rectangulo.perimetro())
print(mi_rectangulo)
```

```python
#Ejercicio 8
class Reproductor:
    def __init__(self):
        self.playlist=[]
        self.posicion=0

    def agregar_cancion(self, cancion:str)->None:
        self.playlist.append(cancion)

    def cancion_actual(self)->str:
        if not self.playlist:
            return 'No hay canciones'
        return self.playlist[self.posicion]

    def siguiente_cancion(self)->str:
        self.posicion+=1
        if self.posicion<len(self.playlist):
            cancion_actual= self.cancion_actual()
            return self.cancion_actual

        self.posicion-=1
        return self.cancion_actual()
    
    def anterior_cancion(self)->str:
        self.posicion-=1
        if self.posicion>=0:
            return self.cancion_actual()
        
        self.posicion+=1
        return self.cancion_actual()

mi_reproductor=Reproductor()
mi_reproductor.agregar_cancion('Mary Lou')
mi_reproductor.siguiente_cancion()
```

```python
#Ejercicio 9
class Motor:
    def __init__(self, tipo:str, caballos:float)->None: #entonces, no habra return al final

        self.caballos=caballos
        self.tipo=tipo
        self.encendido=False
        #Cuando no va return, se sobreentiende que es None


    def arrancar(self):
        if self.encendido:
            return 'EL motor ya esta encendido'

        self.encendido=True
        return f'El motor de {self.tipo} de {self.caballos} hp se ha encendido'
        
class Vehiculo:
    def __init__(self, marca:str, motor:Motor)->None:

        self.marca=marca
        self.motor=motor

    def arrancar(self)->str:
        return self.motor.arrancar()

motor=Motor('petroleo',100)
vehiculo=Vehiculo('Toyota',motor)

print(vehiculo.arrancar())
print(vehiculo.arrancar())
```

```python
#Ejercicio 10
class Libro:
    def __init__(self, titulo:str,autor:str)->None:
        self.titulo=titulo
        self.autor=autor
        self.disponible=True

class Usuario:
    def __init__(self, nombre=str)->None:
        self.nombre=nombre
        self.libros_prestados=[]
        
class Biblioteca:
    def __init__(self):
        self.libros=[]

    def agregar_libro(self,libro:Libro)->None:
        self.libros.append(libro)

    def buscar_libro(self,titulo:str)->Libro | None:
        for libro in self.libros:
            if libro.titulo.lower()==titulo.lower():
                return libro

        return None

    def prestar(self, titulo:str,usuario:Usuario)->str:
        libro=self.buscar_libro(titulo)

        if libro is None:
            return 'El libro no se encuentra en la biblioteca'

        if not libro.disponible:
            return 'El libro ya fue prestado'
        libro.disponible=False
        usuario.libros_prestados.append(libro)

        return f'Libro prestado: {libro.titulo} al usuario: {usuario.nombre}'

    def devolver(self,titulo:str, usuario:str)->str:
        libro=self.buscar_libro(titulo)
        if libro is None:
            return 'No existe el libro'
        
        if libro not in usuario.libros_prestados:
            return f'{usuario.nombre} no tiene prestado el libro: {titulo}'

        usuario.libros_prestados.remove(libro)
        libro.disponible=True
        return f'El libro {titulo} fue devuelto por el usuario {usuario.nombre}'

    def catalogo(self)->None:
        for libro in self.libros:
            print(f'Titulo:{libro.titulo} | Autor:{libro.autor}')

biblioteca=Biblioteca()
libro1=Libro('El caballero carmelo','Abraham Baldelomar')
libro2=Libro('Rebelion en la granja', 'George Orwell')
libro3=Libro('Quien mato a Palomino Molero', 'Mario vargas')

biblioteca.agregar_libro(libro1)
biblioteca.agregar_libro(libro2)
biblioteca.agregar_libro(libro3)

biblioteca.catalogo()

usuario1=Usuario('Pedro')
usuario2=Usuario('Juan')
usuario3=Usuario('Sara')

print('='*20)

print(
    biblioteca.prestar('Rebelion en la granja', usuario1)
    )
print('='*20)

biblioteca.catalogo()

for libro in usuario1.libros_prestados:
    print(libro.titulo)

print(
    biblioteca.prestar('Metamorfosis', usuario2)
    )
print('='*20)

print(
    biblioteca.prestar('Rebelion en la granja', usuario2)
    )
print('='*20)

print(
    biblioteca.prestar('El caballero carmelo', usuario3)
    )
print('='*20)
```
