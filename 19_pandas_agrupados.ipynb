```python
import pandas as pd
ventas=pd.read_csv(
    '/Users/HP/Documents/Data_Science/01-python-intro/ventas.csv',
    dtype={'precio_unitario':'float64'}
)    
ventas['importe']=ventas['unidades']*ventas['precio_unitario']
ventas
```

```python
ventas.groupby('region')['importe'].sum()
```

```python
ventas[ventas['region']=='Sur']['importe'].sum()
```

```python
ventas.groupby('region')['importe'].agg(
    importe_total='sum',
    importe_promedio='mean',
    total_registros='count'
    )
```

```python
ventas.groupby('region').agg({
    'importe':['sum','mean','max'],
    'unidades':['sum'],
    'producto':['nunique']
})
```
