```python
import numpy as np
from matplotlib import pyplot as plt
```

```python
x=np.linspace(0,5,11)
y1=x**2
y2=x**3
plt.subplot(1,2,1)
plt.plot(x,y1,'bo-')
plt.title('funcion cuadratica')

plt.subplot(1,2,2)
plt.plot(x,y2,'go-')
plt.title('funcion cubica')

plt.tight_layout(pad=0.01, h_pad=0.5,w_pad=10)
plt.show()
```

```python
x=np.linspace(0,5,11)
y1=x**2
y2=x**3
frutas=['pera','manzana','platano']
porcentaje=[25,35,40]

categorias= ['A', 'B','C']
valores=[12,34,67]

plt.figure(figsize=(12,6))

plt.subplot(2,2,1)
plt.plot(x,y1,'bo-')
plt.title('funcion cuadratica')

plt.subplot(2,2,2)
plt.plot(x,y2,'go-')
plt.title('funcion cubica')

plt.subplot(2,2,3)
plt.pie(porcentaje, labels=frutas,autopct='%1.1f%%')
plt.title('Grafico de pastel')

plt.subplot(2,2,4)
plt.bar(categorias,valores,color=['red','black','blue'])
plt.xlabel('categorias')
plt.ylabel('valores')
plt.title('Grafico de barras')


plt.tight_layout(pad=0.01, h_pad=0.5,w_pad=10)
plt.show()
```
