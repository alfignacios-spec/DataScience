```python
x = np.linspace(0, 5, 11)
y1 = x ** 2
y2 = x ** 3
```

```python
def grafico_cuadratico(ax, x, y):
    ax.plot(x, y, 'ro--')
    ax.set_title('Función cuadrática')
    ax.grid(True, linestyle='--', axis='both', gapcolor='green', linewidth=2)
    ax.set_xlabel('Input')
    ax.set_ylabel('Output')


def grafico_cubico(ax, x, y):
    ax.plot(x, y, 'bo-')
    ax.set_title('Función cúbica')
    ax.grid(True)

def grafico_pastel(ax, pct, labels):
    ax.pie(pct, labels=labels, autopct='%1.1f%%')
    ax.set_title('Gráfico Pastel')


dashboard, axes = plt.subplots(1, 2, figsize=(12, 5))

grafico_cuadratico(axes[0], x, y1)
grafico_cubico(axes[1], x, y2)

dashboard.suptitle('Dashboard')
plt.tight_layout()
plt.show()
```

```python
frutas = ['Manzanas', 'Plátanos', 'Peras']
porcentajes = [30, 50, 20]


dashboard, axes = plt.subplots(1, 3, figsize=(12, 5))

grafico_cuadratico(axes[0], x, y1)
grafico_pastel(axes[1], porcentajes, frutas)
grafico_cubico(axes[2], x, y2)


dashboard.suptitle('Dashboard')
plt.tight_layout()
plt.show()
```

```python
import numpy as np
plt.style.use('Solarize_Light2')

x=np.linspace(1,20,100)
dashboard, axes =plt.subplots(2,4,figsize=(12,6))
axes[0,0].plot(x,np.cos(x))

axes[0,2].plot(x,np.tanh(x),'r--')

axes[1,3].plot(x,np.sin(x),'r-.',label='f(x)=sin(x)')
axes[1,3].set_title('Onda sinosoidal')
axes[1,3].grid(True,color='blue', linestyle='--',alpha=0.4)
dashboard.tight_layout()
plt.show()
```
