```python
class Estudiante:
    def __init__(
        self,
        nombre: str,
        edad: int,
        ciudad: str,
        nota: float
    ) -> None:
        self.nombre = nombre
        self.edad = edad
        self.ciudad = ciudad
        self.nota = nota


class Aula:
    def __init__(
        self,
        estudiantes: list[str]
    ) -> None:
        self.estudiantes: list[Estudiante] = []

        for estudiante in estudiantes:
            info = [item.strip() for item in estudiante.split(",")]
            estudiante = Estudiante(
                nombre=info[0],
                edad=int(info[1]),
                ciudad=info[2],
                nota=float(info[3])
            )
            self.estudiantes.append(estudiante)

    def __len__(self) -> int:
        return len(self.estudiantes)

    def filtro_por_ciudad(self, ciudad: str) -> None:
        for estudiante in self.estudiantes:
            if estudiante.ciudad.lower() == ciudad:
                print(estudiante.nombre)

    def get_aprobados(self, min_nota: int = 13) -> list[str]:
        estudiantes = []
        for estudiante in self.estudiantes:
            if estudiante.nota >= min_nota:
                estudiantes.append(estudiante.nombre)

        return estudiantes

    def get_num_aprobados(self, min_nota: int = 13) -> int:
        return len(self.get_aprobados(min_nota))

    def export_alumnos(self, ruta: str) -> None:
        with open(ruta, "w") as f:
            f.write("nombre, edad, ciudad, nota\n")

        for estudiante in self.estudiantes:
            with open(ruta, "a") as f:
                f.write(f"{estudiante.nombre}, {estudiante.edad}, {estudiante.ciudad}, {estudiante.nota}\n")

aula = Aula(contenido)
```

```python
class Venta:
    
    def __init__(
        self,
        id: str,
        producto: str,
        talla: str,
        color: str,
        cliente: str,
        ciudad: str,
        cantidad: str,
        precio_unitario: str,
        metodo_pago: str,
        estado: str
    ) -> None:
        self.id = id.strip()
        self.producto = producto.strip()
        self.talla = talla.strip()
        self.color = color.strip()
        self.cliente = cliente.strip()
        self.ciudad = ciudad.strip()
        self.cantidad = int(cantidad.strip())
        self.precio_unitario = float(precio_unitario.strip())
        self.metodo_pago = metodo_pago.strip()
        self.estado = estado.strip()

    def calcular_total(self) -> float:
        return self.cantidad * self.precio_unitario

    def esta_entregada(self) -> bool:
        if self.estado.lower() == "entregado":
            return True


class RegistroVentas:
    def __init__(self, ventas: list[str]) -> None:
        self.ventas: list[Venta] = []
        for venta in ventas:
            info = venta.split(",")
            venta = Venta(*info)
            self.ventas.append(venta)

        # for venta in self.ventas:
        #     print(venta.id)

    def contar_por_estado(self) -> dict[str, int]:
        estados = [venta.estado for venta in self.ventas]
        estados_unicos = set(estados)
        conteos = {}
        for estado in estados_unicos:
            conteos[estado] = estados.count(estado)
        return conteos

    def ingreso_total_ventas_entregadas(self) -> float:
        ingreso_total = 0
        for venta in self.ventas:
            if venta.esta_entregada():
                ingreso_total += venta.calcular_total()

        return ingreso_total

    def get_cantidad_por_producto(self) -> dict[str, int]:
        cantidades = {}
        for venta in self.ventas:
            if venta.producto not in cantidades:
                cantidades[venta.producto] = 0

            cantidades[venta.producto] += venta.cantidad

        return cantidades

    def obtener_producto_mas_vendido(self) -> tuple[str, int]:
        cantidades = self.get_cantidad_por_producto()
        producto = max(cantidades, key=cantidades.get)
        return producto, cantidades[producto]

    def contar_ventas_por_ciudad(self, ciudad: str) -> int:
        cantidad_ventas = 0
        for venta in self.ventas:
            if venta.ciudad.lower() == ciudad.lower():
                cantidad_ventas += 1

        return cantidad_ventas

    def obtener_metodo_pago_mas_usado(self) -> tuple[str, int]:
        metodos_pago = {}
        for venta in self.ventas:
            if venta.metodo_pago not in metodos_pago:
                metodos_pago[venta.metodo_pago] = 0

            metodos_pago[venta.metodo_pago] += 1

        metodo_mas_usado = max(
            metodos_pago,
            key=metodos_pago.get
        )

        return metodo_mas_usado, metodos_pago[metodo_mas_usado]


registro_ventas = RegistroVentas(contenido)
print(registro_ventas.contar_por_estado())
print(registro_ventas.ingreso_total_ventas_entregadas())
print(registro_ventas.get_cantidad_por_producto())
print(registro_ventas.obtener_producto_mas_vendido())
print(registro_ventas.contar_ventas_por_ciudad("Arequipa"))
print(registro_ventas.obtener_metodo_pago_mas_usado())
```
